###  \[Barplot o grafico de barras\]

Es un grafico donde represento visualmente en rectangulos \(barras\) para comparar cantidades en distintas categorias,en este caso la cantidad de delitos por barrio\.

```
top_barrios = (delitos.groupby("barrio")["cantidad"].sum().sort_values(ascending=False).head(15)
)

plt.figure(figsize=(10, 7))
sns.barplot(
    x=top_barrios.values,
    y=top_barrios.index,
    palette="Reds_r"
)
plt.title("Top 15 barrios con más delitos (CABA 2022)",
          fontsize=14, fontweight="bold")
plt.xlabel("Cantidad de delitos")
plt.ylabel("Barrio")
plt.tight_layout()
plt.savefig("top_barrios.png", dpi=150)
plt.show()
```

###  \[heatmap o mapa de calor\]

El mapa de calor me representa visualmente  datos que utiliza una escala de colores para mostrar la concentracion o densidad de algun evento\.

```
orden_dias = ["LUN", "MAR", "MIE", "JUE", "VIE", "SAB", "DOM"]
delitos["dia"] = pd.Categorical(delitos["dia"], categories=orden_dias, ordered=True)


pivot = delitos.pivot_table(
    index="dia",
    columns="franja",
    values="cantidad",
    aggfunc="sum",
    fill_value=0)

plt.figure(figsize=(14, 6))
sns.heatmap(
    pivot,
    cmap="YlOrRd",
    linewidths=0.3,
    linecolor="white",
    cbar_kws={"label": "Cantidad de delitos"}
)
plt.title("Delitos por día de la semana y franja horaria (CABA 2022)",
          fontsize=14, fontweight="bold")
plt.xlabel("Franja horaria (hora del día)")
plt.ylabel("Día de la semana")
plt.tight_layout()
plt.savefig("heatmap_temporal.png", dpi=150)
plt.show()
```

📌 Vemos que los dias con mas delitos son de lunes a viernes en una franja horaria que va desde las 7 de la mañana a las 21 horas esto me hace interpretar que la mayor cantidad de delitos se concentra en dias laborales y en una franja horaria laboral\.

# Valores atipicos y inconsistentes 

posibles valores atípicos cuartiles

```
for columna in ["price", "mileage", "year"]:
    Q1 = autos[columna].quantile(0.25)
    Q3 = autos[columna].quantile(0.75)

    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    atipicos = autos[
        (autos[columna] < limite_inferior) |
        (autos[columna] > limite_superior)
    ]

    print("\nVariable:", columna)
    print("Límite inferior:", limite_inferior)
    print("Límite superior:", limite_superior)
    print("Cantidad de posibles atípicos:", len(atipicos))
```

Valores que podrían ser inconsistentes

```
print(autos.nlargest(10, "price")[
    ["model", "year", "price", "mileage"]
])

print(autos.nlargest(10, "mileage")[
    ["model", "year", "price", "mileage"]
])


print("Motor igual a cero:", (autos["engineSize"] == 0).sum())
print("MPG igual a 470.8:", (autos["mpg"] == 470.8).sum())
print("Impuesto igual a cero:", (autos["tax"] == 0).sum())
```

Corrección y profundización del análisis de faltantes \(info\(\)\):
Tras auditar la distribución de valores nulos por tipo de desastre, se confirmó la hipótesis de nulos estructurales, pero se identificó una inconsistencia relevante en el dataset de origen: Distribución de variables de magnitud sísmica: 
Los 1\.376 registros no nulos de magnitud\_richter, profundidad\_km y genero\_tsunami se concentran en su totalidad dentro de los 1\.636 eventos etiquetados como Deslizamiento de tierra, registrando 0 valores no nulos en la categoría Terremoto\.Distribución de variables generales: Las variables estacion y severidad \(3\.917 registros no nulos\) están presentes en el 100\\% de las demás categorías \(incluyendo Terremoto con 1\.190 casos\) y en los 260 eventos restantes de Deslizamiento de tierra \(1\.376 \+ 260 = 1\.636\)\.Conclusión del análisis: Los valores faltantes no corresponden a una omisión aleatoria en la recolección, sino a un solapamiento o error de clasificación en la variable tipo\_desastre dentro de la fuente original, donde mediciones de origen telúrico fueron asociadas a eventos de deslizamiento\. Se decide conservar el dataset en su estado original documentando este hallazgo como parte del diagnóstico de calidad de datos\.

"Convendría verificar la procedencia y consistencia de cifras tan elevadas como las muertes acumuladas por deslizamientos\."

```
print("--- Muertes acumuladas por tipo de desastre ---")
print(desastre.groupby('tipo_desastre')['muertes'].sum().sort_values(ascending=False))

desliz_sismos = desastre[(desastre['tipo_desastre'] == 'Deslizamiento de tierra') & (desastre['magnitud_richter'].notnull())]
desliz_puros = desastre[(desastre['tipo_desastre'] == 'Deslizamiento de tierra') & (desastre['magnitud_richter'].isnull())]

print("\n--- Desglose de muertes en 'Deslizamiento de tierra' ---")
print(f"Eventos totales: {len(desastre[desastre['tipo_desastre'] == 'Deslizamiento de tierra'])}")
print(f"Muertes provienen de eventos con Richter (Sismos mal etiquetados): {desliz_sismos['muertes'].sum():,}")
print(f"Muertes provienen de eventos sin Richter (Deslizamientos puros): {desliz_puros['muertes'].sum():,}")
```

La cifra de 429\.282 muertes acumuladas por deslizamientos es engañosa y se explica por un error de etiquetado en la fuente original de datos:
\* El 92,1% de los fallecidos son en realidad víctimas de sismos: De las 429\.282 muertes atribuidas a Deslizamiento de tierra, 395\.598 corresponden a terremotos que fueron mal clasificados dentro de esta categoría \(los 1\.376 eventos con escala Richter\)\.
\* Impacto real de los deslizamientos: Los deslizamientos puros \(los 260 eventos reales sin escala Richter\) solo suman 33\.684 muertes, un valor coherente con la escala media de este fenómeno\.
\* Efecto en el ranking de mortalidad: Si se corrigiera el etiquetado, la categoría Terremoto pasaría a liderar el dataset con 539\.058 muertes reales \(143\.460 \+ 395\.598\), confirmando que la aparente dominancia de los deslizamientos es únicamente un espejismo provocado por la distorsión del dataset

###  \[Hipótesis 2\]

La disatncia del hogar \(DistanceFromHome\)es un factor determinante en la rotacion de personal \(Attrition\_numerico\)\.
\(No Cumplida\)

Visualizo los valores unicos de la columna Distancia al trabajo \(DistanceFromHome\)

Las distancias se categorizan en 4 rangos cualitativos para facilitar su analisis:" Cerca \(1\-5 km\)", "Media \(6\-10 km\)", "Lejana \(11\-20 km\)", "Muy lejana \(\>20 km\)"

```
bins = [0, 5, 10, 20, np.inf]
labels = ["Cerca (1-5 km)", "Media (6-10 km)", "Lejana (11-20 km)", "Muy lejana (>20 km)"]

df_limpio["Rango_Distancia"] = pd.cut(
    df_limpio["DistanceFromHome"], bins=bins, labels=labels, right=True
)

print("Cantidad de empleados por rango de distancia:")
print(df_limpio["Rango_Distancia"].value_counts().sort_index())
```

Visualizacion en grafico de barras agrupadas\. Muestra la cantidad de empleados en cada rango de distancia al trabajo, divididos segun si permanecieron o dejaron la empresa\. 

```
plt.figure(figsize=(8, 5))
sns.countplot(data=df_limpio, x='Rango_Distancia', hue='Attrition')
plt.title('H2: Rotación de Empleados según Rango de Distancia')
plt.xlabel('Rango de Distancia')
plt.ylabel('Cantidad de Empleados')
plt.show()
```

Distribución proporcional constante: Al analizar la distribución de la rotación en los diferentes rangos de distancia \(desde Cerca hasta Muy lejana\), se observa que la proporción de empleados que deciden dejar la empresa \(Yes\) en relación con los que se quedan \(No\) se mantiene estable\. Si la distancia fuera un factor determinante, se esperaría un incremento significativo y desproporcionado en la tasa de abandono a medida que aumentan los kilómetros de viaje\.

La Hipótesis no se cumple

###  \[Hipótesis 3\]

Un salario mas alto garantiza un mayor nivel de satisfaccion laboral\(JobSatisfaction\)\. \(No Cumplida\)

```
orden_salarios = ["Upto 5k", "5k-10k", "10k-15k", "15k+"]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(
    data=df_limpio,
    x="JobSatisfaction",
    y="MonthlyIncome",
    color="lightblue",
    ax=axes[0]
)
axes[0].set_title("Ingreso mensual según satisfacción laboral")
axes[0].set_xlabel("Satisfacción laboral (1 = baja, 4 = alta)")
axes[0].set_ylabel("Ingreso mensual ($)")

sns.pointplot(
    data=df_limpio,
    x="SalarySlab",
    y="JobSatisfaction",
    order=orden_salarios,
    estimator=np.median,
    errorbar=None,
    color="darkorange",
    ax=axes[1]
)
axes[1].set_title("Satisfacción mediana por rango salarial")
axes[1].set_xlabel("Rango de ingreso mensual")
axes[1].set_ylabel("Satisfacción laboral mediana")
axes[1].tick_params(axis="x", rotation=20)
axes[1].set_ylim(1, 4)

plt.tight_layout()
plt.show()


print("Ingreso mediano por nivel de satisfacción:")
print(df_limpio.groupby("JobSatisfaction", observed=False)["MonthlyIncome"].median())

print("\nSatisfacción mediana por rango salarial:")
print(df_limpio.groupby("SalarySlab", observed=False)["JobSatisfaction"].median())

rho = df_limpio["MonthlyIncome"].corr(df_limpio["JobSatisfaction"], method="spearman")
print(f"\nCorrelación de Spearman (MonthlyIncome vs JobSatisfaction): {rho:.3f}")
```

A través del análisis gráfico y numérico, se observa que la satisfacción laboral mediana se mantiene prácticamente idéntica \(3 sobre 4\) en todos los rangos salariales \(SalarySlab\)\. Asimismo, la correlación de Spearman cercana a 0 confirma que percibir un salario más elevado no garantiza un mayor nivel de satisfacción laboral en esta organización

La Hipótesis no se cumple

###  \[Hipótesis 4\]

A mayor antigüedad del empleado en la empresa \(YearsAtCompany\), mayor es el tiempo transcurrido desde su último ascenso \(YearsSinceLastPromotion\)\.

```
bins = [-1, 2, 5, 10, 20, np.inf]
labels = ["0-2 años", "3-5 años", "6-10 años", "11-20 años", "+20 años"]
df_limpio["Rango_Antiguedad"] = pd.cut(
    df_limpio["YearsAtCompany"], bins=bins, labels=labels, include_lowest=True
)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.regplot(
    data=df_limpio,
    x="YearsAtCompany",
    y="YearsSinceLastPromotion",
    x_jitter=0.3,
    y_jitter=0.3,
    scatter_kws={"alpha": 0.25, "s": 15},
    line_kws={"color": "red"},
    ax=axes[0]
)
axes[0].set_title("Antigüedad vs. Años desde el último ascenso", fontweight="bold")
axes[0].set_xlabel("Años en la empresa (YearsAtCompany)")
axes[0].set_ylabel("Años desde el último ascenso (YearsSinceLastPromotion)")

sns.barplot(
    data=df_limpio,
    x="Rango_Antiguedad",
    y="YearsSinceLastPromotion",
    color="salmon",
    errorbar=None,
    ax=axes[1]
)
axes[1].set_title("Años promedio sin ascenso según antigüedad", fontweight="bold")
axes[1].set_xlabel("Rango de antigüedad")
axes[1].set_ylabel("Promedio de años desde el último ascenso")

plt.tight_layout()
plt.show()

print("Promedio de años sin ascenso por rango de antigüedad:")
print(
    df_limpio.groupby("Rango_Antiguedad", observed=True)["YearsSinceLastPromotion"]
    .agg(["count", "mean"])
    .round(2)
)

r = df_limpio["YearsAtCompany"].corr(df_limpio["YearsSinceLastPromotion"])
rho = df_limpio["YearsAtCompany"].corr(
    df_limpio["YearsSinceLastPromotion"], method="spearman"
)
print(f"\nCorrelación de Pearson: {r:.3f} | Spearman: {rho:.3f}")
```

El análisis muestra una correlación positiva considerable entre los años en la empresa \(YearsAtCompany\) y el tiempo transcurrido desde el último ascenso \(YearsSinceLastPromotion\)\. El promedio de años sin ascenso crece sustancialmente a medida que los empleados pertenecen a rangos de mayor antigüedad, confirmando la hipótesis planteada

La Hipótesis se cumple
