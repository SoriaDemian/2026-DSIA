###  \[Barplot o grafico de barras\]

Es un grafico donde represento visualmente en rectangulos \(barras\) para comparar cantidades en distintas categorias,en este caso la cantidad de delitos por barrio\.

```
top_barrios = (delitos.groupby("barrio")["cantidad"].sum().sort_values(ascending=False).head(15)
)

plt.figure(figsize=(10, 7))
sns.barplot(
    x=top_barrios.values,
    y=top_barrios.index,
    palette="Reds_r"
)
plt.title("Top 15 barrios con más delitos (CABA 2022)",
          fontsize=14, fontweight="bold")
plt.xlabel("Cantidad de delitos")
plt.ylabel("Barrio")
plt.tight_layout()
plt.savefig("top_barrios.png", dpi=150)
plt.show()
```

###  \[heatmap o mapa de calor\]

El mapa de calor me representa visualmente  datos que utiliza una escala de colores para mostrar la concentracion o densidad de algun evento\.

```
orden_dias = ["LUN", "MAR", "MIE", "JUE", "VIE", "SAB", "DOM"]
delitos["dia"] = pd.Categorical(delitos["dia"], categories=orden_dias, ordered=True)


pivot = delitos.pivot_table(
    index="dia",
    columns="franja",
    values="cantidad",
    aggfunc="sum",
    fill_value=0)

plt.figure(figsize=(14, 6))
sns.heatmap(
    pivot,
    cmap="YlOrRd",
    linewidths=0.3,
    linecolor="white",
    cbar_kws={"label": "Cantidad de delitos"}
)
plt.title("Delitos por día de la semana y franja horaria (CABA 2022)",
          fontsize=14, fontweight="bold")
plt.xlabel("Franja horaria (hora del día)")
plt.ylabel("Día de la semana")
plt.tight_layout()
plt.savefig("heatmap_temporal.png", dpi=150)
plt.show()
```

📌 Vemos que los dias con mas delitos son de lunes a viernes en una franja horaria que va desde las 7 de la mañana a las 21 horas esto me hace interpretar que la mayor cantidad de delitos se concentra en dias laborales y en una franja horaria laboral\.

# Valores atipicos y inconsistentes 

posibles valores atípicos cuartiles

```
for columna in ["price", "mileage", "year"]:
    Q1 = autos[columna].quantile(0.25)
    Q3 = autos[columna].quantile(0.75)

    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    atipicos = autos[
        (autos[columna] < limite_inferior) |
        (autos[columna] > limite_superior)
    ]

    print("\nVariable:", columna)
    print("Límite inferior:", limite_inferior)
    print("Límite superior:", limite_superior)
    print("Cantidad de posibles atípicos:", len(atipicos))
```

Valores que podrían ser inconsistentes

```
print(autos.nlargest(10, "price")[
    ["model", "year", "price", "mileage"]
])

print(autos.nlargest(10, "mileage")[
    ["model", "year", "price", "mileage"]
])


print("Motor igual a cero:", (autos["engineSize"] == 0).sum())
print("MPG igual a 470.8:", (autos["mpg"] == 470.8).sum())
print("Impuesto igual a cero:", (autos["tax"] == 0).sum())
```

Corrección y profundización del análisis de faltantes \(info\(\)\):
Tras auditar la distribución de valores nulos por tipo de desastre, se confirmó la hipótesis de nulos estructurales, pero se identificó una inconsistencia relevante en el dataset de origen: Distribución de variables de magnitud sísmica: 
Los 1\.376 registros no nulos de magnitud\_richter, profundidad\_km y genero\_tsunami se concentran en su totalidad dentro de los 1\.636 eventos etiquetados como Deslizamiento de tierra, registrando 0 valores no nulos en la categoría Terremoto\.Distribución de variables generales: Las variables estacion y severidad \(3\.917 registros no nulos\) están presentes en el 100\\% de las demás categorías \(incluyendo Terremoto con 1\.190 casos\) y en los 260 eventos restantes de Deslizamiento de tierra \(1\.376 \+ 260 = 1\.636\)\.Conclusión del análisis: Los valores faltantes no corresponden a una omisión aleatoria en la recolección, sino a un solapamiento o error de clasificación en la variable tipo\_desastre dentro de la fuente original, donde mediciones de origen telúrico fueron asociadas a eventos de deslizamiento\. Se decide conservar el dataset en su estado original documentando este hallazgo como parte del diagnóstico de calidad de datos\.

"Convendría verificar la procedencia y consistencia de cifras tan elevadas como las muertes acumuladas por deslizamientos\."

```
print("--- Muertes acumuladas por tipo de desastre ---")
print(desastre.groupby('tipo_desastre')['muertes'].sum().sort_values(ascending=False))

desliz_sismos = desastre[(desastre['tipo_desastre'] == 'Deslizamiento de tierra') & (desastre['magnitud_richter'].notnull())]
desliz_puros = desastre[(desastre['tipo_desastre'] == 'Deslizamiento de tierra') & (desastre['magnitud_richter'].isnull())]

print("\n--- Desglose de muertes en 'Deslizamiento de tierra' ---")
print(f"Eventos totales: {len(desastre[desastre['tipo_desastre'] == 'Deslizamiento de tierra'])}")
print(f"Muertes provienen de eventos con Richter (Sismos mal etiquetados): {desliz_sismos['muertes'].sum():,}")
print(f"Muertes provienen de eventos sin Richter (Deslizamientos puros): {desliz_puros['muertes'].sum():,}")
```

La cifra de 429\.282 muertes acumuladas por deslizamientos es engañosa y se explica por un error de etiquetado en la fuente original de datos:
\* El 92,1% de los fallecidos son en realidad víctimas de sismos: De las 429\.282 muertes atribuidas a Deslizamiento de tierra, 395\.598 corresponden a terremotos que fueron mal clasificados dentro de esta categoría \(los 1\.376 eventos con escala Richter\)\.
\* Impacto real de los deslizamientos: Los deslizamientos puros \(los 260 eventos reales sin escala Richter\) solo suman 33\.684 muertes, un valor coherente con la escala media de este fenómeno\.
\* Efecto en el ranking de mortalidad: Si se corrigiera el etiquetado, la categoría Terremoto pasaría a liderar el dataset con 539\.058 muertes reales \(143\.460 \+ 395\.598\), confirmando que la aparente dominancia de los deslizamientos es únicamente un espejismo provocado por la distorsión del dataset

### 4\. Aplicación de modelos de clasificación

### Preparación de los datos y división en entrenamiento y prueba

Los modelos necesitan variables numéricas, así que convertimos las categóricas con one\-hot encoding y definimos Attrition como variable objetivo \(1 = se fue\)\. Separamos 80% para entrenar y 20% para evaluar, manteniendo la proporción de renuncias en ambos conjuntos \(stratify\), porque las clases están desbalanceadas\.

```

if "Attrition_numerico" in df_limpio.columns:
    y = df_limpio["Attrition_numerico"]
else:
    y = (df_limpio["Attrition"] == "Yes").astype(int)

# Columnas que NO deben usarse como predictoras:
# la respuesta, el identificador y las versiones agrupadas de otras variables
cols_fuera = ["Attrition", "Attrition_numerico", "EmpID", "AgeGroup", "SalarySlab"]
X = df_limpio.drop(columns=[c for c in cols_fuera if c in df_limpio.columns])

# Convertir variables categóricas a numéricas
X = pd.get_dummies(X, drop_first=True)

# División 80/20 estratificada
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)

print("Variables predictoras:", X.shape[1])
print("Proporción de renuncias:", round(y.mean(), 3))
print("Train:", X_train.shape, "| Test:", X_test.shape)
```

Con el dataset original quedan 45 variables predictoras y 3000 registros\. Solo el 11,9% de los empleados renunció, por lo que un modelo que siempre responda “No” ya acertaría ~88%; por eso más adelante no juzgamos los modelos solo por la accuracy\.

### Entrenamiento de los tres modelos

Entrenamos tres modelos de complejidad creciente: Regresión Logística \(lineal e interpretable\), Árbol de Decisión y Random Forest \(conjunto de árboles\)\. A los tres les indicamos que den más peso a la clase minoritaria \(class\_weight\), para que no ignoren a quienes renuncian\.

```
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

modelos = {
    "Regresión Logística": Pipeline([
        ("escala", StandardScaler()),
        ("modelo", LogisticRegression(max_iter=1000, class_weight="balanced", random_state=42))]),
    "Árbol de Decisión": DecisionTreeClassifier(
        max_depth=5, class_weight="balanced", random_state=42),
    "Random Forest": RandomForestClassifier(
        n_estimators=300, max_depth=6, min_samples_leaf=10,
        class_weight="balanced_subsample", random_state=42, n_jobs=-1),
}

for nombre, modelo in modelos.items():
    modelo.fit(X_train, y_train)
    print(f"{nombre} entrenado")
```

Los tres modelos quedaron entrenados con los mismos datos, lo que permite compararlos de forma justa\. La regresión logística necesita las variables escaladas \(por eso lleva StandardScaler\), mientras que los modelos de árboles no\. En el Random Forest limitamos la profundidad y el tamaño de las hojas: sin eso, en las pruebas el modelo terminaba prediciendo “No” para todos los empleados\.

```
log = modelos["Regresión Logística"].named_steps["modelo"]
print("class_weight:", log.class_weight)
print("Empleados que predice como 'Renuncia':", modelos["Regresión Logística"].predict(X_test).sum())
```

### 5\. Métricas de evaluación

### Cálculo de métricas y matrices de confusión

Evaluamos cada modelo sobre el 20% que no vio al entrenar\. Como las clases están desbalanceadas, miramos precision, recall, F1 y AUC\-ROC además de la accuracy\. Sumamos una validación cruzada de 5 particiones del F1 para comprobar que el resultado no depende de una sola división de los datos\.

```
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score, confusion_matrix,
                             ConfusionMatrixDisplay)
from sklearn.model_selection import cross_val_score, StratifiedKFold

filas = []
fig, ejes = plt.subplots(1, 3, figsize=(15, 4))
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

for eje, (nombre, modelo) in zip(ejes, modelos.items()):
    pred = modelo.predict(X_test)
    proba = modelo.predict_proba(X_test)[:, 1]
    filas.append({
        "Modelo": nombre,
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred),
        "Recall": recall_score(y_test, pred),
        "F1": f1_score(y_test, pred),
        "AUC-ROC": roc_auc_score(y_test, proba),
        "F1 (validación cruzada)": cross_val_score(modelo, X, y, cv=cv, scoring="f1").mean(),
    })
    ConfusionMatrixDisplay(confusion_matrix(y_test, pred),
                           display_labels=["Se queda", "Renuncia"]).plot(ax=eje, colorbar=False)
    eje.set_title(nombre)

plt.tight_layout()
plt.show()

resultados = pd.DataFrame(filas).round(3)
display(resultados)
```

Los resultados obtenidos fueron:

Modelo	                      Accuracy	  Precision	Recall	   F1	     AUC\-ROC	  F1 \(CV\)
Regresión Logística	0,648	    0,182	        0,563	0,275  	0,642	  0,242
Árbol de Decisión	        0,585	    0,158	        0,577	0,248	0,587	  0,225
Random Forest	        0,817	    0,247	        0,268	0,257	0,588	  0,231

El Random Forest tiene la accuracy más alta, pero engaña: acierta más porque predice “Se queda” casi siempre, y solo detecta cerca de 19 de las 71 renuncias del conjunto de prueba \(recall 0,268\)\. La Regresión Logística y el Árbol de Decisión detectan más de la mitad de las renuncias \(recall de 0,563 y 0,577\), a costa de más falsas alarmas, lo que explica una precisión baja en los tres modelos \(entre 0,158 y 0,247\)\.

### 6\. Selección del modelo más indicado

Elegimos el modelo según las métricas que importan para este problema: detectar a quienes podrían renunciar \(recall\) y equilibrarlo con la precisión \(F1 y AUC\)\. La accuracy se descarta como criterio por el desbalance de clases\. Después miramos qué variables usa el modelo elegido, para relacionarlo con las hipótesis\.

```
# Ranking por F1 y AUC
display(resultados.sort_values(["F1", "AUC-ROC"], ascending=False).reset_index(drop=True))

# Coeficientes de la Regresión Logística (variables escaladas)
log = modelos["Regresión Logística"].named_steps["modelo"]
coef = pd.Series(log.coef_[0], index=X.columns).sort_values()

fig, ax = plt.subplots(figsize=(8, 6))
pd.concat([coef.head(8), coef.tail(8)]).plot(kind="barh", ax=ax)
ax.set_title("Variables que más influyen en la renuncia (Regresión Logística)")
ax.set_xlabel("Coeficiente (positivo = más riesgo de renuncia)")
plt.tight_layout()
plt.show()
```

El modelo más indicado es la Regresión Logística: tiene el mayor F1 \(0,275\), el mayor AUC\-ROC \(0,642\) y el mayor F1 en validación cruzada \(0,242\), y su recall de 0,563 es casi igual al del Árbol de Decisión \(0,577\)\. Es decir, detecta más de la mitad de las renuncias reales con el mejor equilibrio entre precisión y recall, y además es el más fácil de explicar\. El Árbol de Decisión detecta apenas más renuncias, pero con peor F1 y peor AUC, y el Random Forest, pese a su accuracy alta, deja sin detectar a la mayoría de quienes renuncian\.

En el gráfico de coeficientes, los valores positivos son factores asociados a mayor riesgo de renuncia y los negativos a menor riesgo\. En nuestra prueba, hacer horas extra \(OverTime\) aparece entre los que aumentan el riesgo, y el nivel del puesto \(JobLevel\), los años totales de experiencia y la satisfacción con el ambiente entre los que lo reducen\.

Como limitación, el desempeño general es moderado \(AUC cercano a 0,64 y precisión baja\)\. Con estas variables el modelo sirve para identificar grupos de riesgo, pero no para predecir con certeza qué empleado se va\.
