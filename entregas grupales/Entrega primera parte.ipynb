2\. Visualizaciones y análisis

###  \[Barplot o grafico de barras\]

Es un grafico donde represento visualmente en rectangulos \(barras\) para comparar cantidades en distintas categorias,en este caso la cantidad de delitos por barrio\.

```
top_barrios = (delitos.groupby("barrio")["cantidad"].sum().sort_values(ascending=False).head(15)
)

plt.figure(figsize=(10, 7))
sns.barplot(
    x=top_barrios.values,
    y=top_barrios.index,
    palette="Reds_r"
)
plt.title("Top 15 barrios con más delitos (CABA 2022)",
          fontsize=14, fontweight="bold")
plt.xlabel("Cantidad de delitos")
plt.ylabel("Barrio")
plt.tight_layout()
plt.savefig("top_barrios.png", dpi=150)
plt.show()
```

📌 En este grafico vemos que en los barrios de palermo, balvanera y flores se concentraron la gran cantidad de delitos y en mataderos y monserrat la menor cantida de delitos en caba en el año 2022\.

###  \[heatmap o mapa de calor\]

El mapa de calor me representa visualmente  datos que utiliza una escala de colores para mostrar la concentracion o densidad de algun evento\.

```
orden_dias = ["LUN", "MAR", "MIE", "JUE", "VIE", "SAB", "DOM"]
delitos["dia"] = pd.Categorical(delitos["dia"], categories=orden_dias, ordered=True)


pivot = delitos.pivot_table(
    index="dia",
    columns="franja",
    values="cantidad",
    aggfunc="sum",
    fill_value=0)

plt.figure(figsize=(14, 6))
sns.heatmap(
    pivot,
    cmap="YlOrRd",
    linewidths=0.3,
    linecolor="white",
    cbar_kws={"label": "Cantidad de delitos"}
)
plt.title("Delitos por día de la semana y franja horaria (CABA 2022)",
          fontsize=14, fontweight="bold")
plt.xlabel("Franja horaria (hora del día)")
plt.ylabel("Día de la semana")
plt.tight_layout()
plt.savefig("heatmap_temporal.png", dpi=150)
plt.show()
```

📌 Vemos que los dias con mas delitos son de lunes a viernes en una franja horaria que va desde las 7 de la mañana a las 21 horas esto me hace interpretar que la mayor cantidad de delitos se concentra en dias laborales y en una franja horaria laboral\.

# 📁 Dataset 2 – \[Ventas\_super\_diciembre\_enero\] \(Integrante: Fernanda Fernandez\)

## 📝 Descripción del Dataset

El Dataset contiene registros de ventas de un supermercado correspondientes a dos fechas puntuales: un día del mes de diciembre y ese mismo día del mes de enero del año siguiente\. El objetivo de tomar estas dos fechas es poder comparar el comportamiento de las ventas entre ambos períodos \(fin de año vs comienzo de año\)\.

## Importación de Librerías

```
import seaborn as sns
import pandas as pd

import matplotlib.pyplot as plt



ventas = pd.read_excel('Ventas_super_diciembre_enero.xlsx')
```

## 1\. Carga y exploración de datos

###  \[Metodo Shape\]

El método shape nos dá un preview del tamaño del data

```
ventas.shape
```

El dataset cuenta con 2880 filas y 11 columnas

###  \[Metodo dtypes \]

El método dtypes nos permite entender con que tipo de datos estámos trabajando

```
ventas.dtypes
```

Las columnas SUCURSAL, EMPRESA, TERMINAL, TRANSACCION y FECHA son numéricas enteras \(int64\); HORA y TIPO\_FAC son texto \(object\); y los importes \(IMP\_TOTAL, IMP\_BRUTO, IMP\_DESC, IMP\_PERCEP\) son numéricos decimales \(float64\)\.

## 2\. Visualizaciones y análisis

### Comparación de ventas: diciembre vs enero

Se realiza un gráfico de conteo para visualizar la cantidad de ventas \(transacciones\) registradas en cada una de las dos fechas analizadas, con el objetivo de comparar el volumen de operaciones entre diciembre y enero\.

```
sns.countplot(x="FECHA", data=ventas)
```

📌 El gráfico muestra la cantidad de transacciones realizadas en cada fecha analizada\. Se observa que el día 21/12/2025 tuvo una mayor cantidad de ventas \(~1670\) en comparación con el mismo día de enero \(~1200\), lo que indica una caída en el volumen de operaciones de diciembre a enero\.

###  \[Importe total por terminal \- Diciembre\]

Se realiza un diagrama de barras para visualizar el importe total vendido por cada terminal durante el mes de diciembre\.

```
diciembre = ventas[ventas['FECHA'] == 21122025]
plt.figure(figsize=(12, 6))
sns.barplot(x="TERMINAL", y="IMP_TOTAL", data=diciembre)
plt.xticks(rotation=45)
plt.show()
```

📌 se observa que el importe total facturado varía notablemente según la terminal, destacándose algunas \(como la 4443 y la 4450\) con montos significativamente más altos que el resto, mientras que otras terminales presentan una facturación baja y más pareja entre sí\.

## 3\. Observaciones generales

El análisis muestra que el día de diciembre tuvo una mayor cantidad de ventas que el mismo día de enero, lo que sugiere una posible estacionalidad asociada a las fiestas de fin de año\. Además, se observó que el importe facturado no se distribuye de forma pareja entre terminales de una misma sucursal: algunas concentran montos mucho más altos que el resto, lo que podría deberse a su ubicación dentro del local, mayor uso o al tipo de ventas que procesan \(por ejemplo, cajas rápidas vs\. cajas de compras grandes\)\.

En cuanto a la calidad de los datos, el dataset se encontraba completo y sin mayores inconvenientes, aunque fue necesario convertir la columna FECHA \(originalmente numérica\) a formato fecha para poder trabajar correctamente con ella\.

Las variables que resultaron más interesantes para el análisis fueron FECHA, TERMINAL e IMP\_TOTAL, ya que permitieron identificar diferencias temporales en el volumen de ventas y detectar qué terminales concentran mayor facturación dentro de la misma sucursal\.

## Importación de Librerías

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1\. Carga y exploración de datos

### Función read\_csv y atributo \.shape

En este primer paso se carga el dataset y se verifica la cantidad total de filas y columnas disponibles en memoria antes de iniciar la exploración\.

```
desastre = pd.read_csv(os.path.join("Dataset_4", "desastres_naturales_2016_2026.csv"))
desastre.shape
```

El dataset contiene 5\.293 filas y 25 columnas, lo que confirma que el dataset se leyó correctamente y está listo para el análisis\.

###  Método \.info\(\)

Se ejecuta el método \.info\(\) para consultar la estructura general del dataset, identificando los tipos de datos de cada columna y la presencia de valores nulos

```
desastre.info()
```

Muestra el detalle de las 25 variables \(textos, números enteros y flotantes\) y podemos detectar que ninguna columna contienen valores nulos que requerirán limpieza\.

## 2\. Visualizaciones y análisis

### Gráfico de barras

Se crea un gráfico de barras horizontales para sumar la cantidad acumulada de la variable muertes por cada tipo de desastre, permitiendo comparar la mortalidad total entre categorías\.

```
sns.barplot(data=desastre, x="muertes", y="tipo_desastre", estimator=sum)

plt.title("Total de muertes por tipo de desastre")
plt.xlabel("Total de muertes")
plt.ylabel("Tipo de desastre")
plt.show()
```

📌 Los deslizamientos de tierra constituyen la principal causa de mortalidad acumulada en el dataset, superando las 400\.000 víctimas\. Le siguen las inundaciones y los terremotos \(ambos por encima de los 140\.000 fallecidos\), mientras que eventos como las avalanchas, sequías e incendios muestran cifras significativamente menores en la escala total\.

###  Gráfico de líneas

Se agrupa la base de datos por año y se cuenta la cantidad total de registros de id\_evento para visualizar la evolución temporal y la frecuencia anual de desastres entre 2016 y 2026\.

```
desastre_anio = desastre.groupby("anio")["id_evento"].count().reset_index()

plt.plot(desastre_anio["anio"], desastre_anio["id_evento"], marker="o", color="purple")

plt.title("Cantidad total de desastres ocurridos por año")
plt.xlabel("Año")
plt.ylabel("Número de Desastres")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

📌 Entre 2016 y 2025 se observa una ocurrencia de desastres muy estable, manteniéndose entre 490 y 540 eventos por año \(con sus puntos más altos en 2020 y 2024\)\. La caída drástica que se aprecia en 2026 responde a que la recolección de datos del año en curso es parcial ya que aun no termina el año\.

## 3\. Observaciones generales

Concentración del impacto humano: Existe una marcada disparidad en la letalidad según el tipo de fenómeno\. Los deslizamientos de tierra, las inundaciones y los terremotos concentran la gran mayoría de las víctimas mortales registradas, mientras que otros eventos \(como sequías o incendios\) presentan cifras de mortalidad directa mucho menores\.
