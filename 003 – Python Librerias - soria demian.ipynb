Ejercicio 2 – Dataset simple con Pandas
Crear un DataFrame de Pandas con la siguiente información:

estudiante	nota
Ana	8
Juan	6
Carla	9
Pedro	5
Lucia	7

Luego realizar las siguientes operaciones:

Mostrar el DataFrame
Calcular el promedio de las notas
Mostrar la nota máxima
Mostrar la nota mínima

```
estudiantes={
    "nombre":["Ana","Juan","Carla","Pedro","Lucia"],
    "nota":[8,6,9,5,7],
}
dataset=pd.DataFrame(estudiantes)
print(dataset)
```

mean hace la media

```
promedio=dataset["nota"].mean() 
print("El promedio de las notas es:",promedio)
```

max es el maximo

```
alta=dataset["nota"].max()
print("La nota mas alta fue:",alta)
```

min es el minimo

```
baja=dataset["nota"].min()
print("La nota mas baja fue:",baja)
```
